```python
import numpy
import matplotlib as mpl
from matplotlib import pyplot as plt
%matplotlib inline

ms_and_clr_by_family = {
    'Bi': ('o','green'),
    'Tl/O': ('D','blue'),
    'Hg/Ba': ('*','red'),
    "Tp": ("x",'purple'),
    "Chain": ("s",'orange'),
    "CuLayer": ("+",'brown'),
    "Other": (".",'black')
}
```

```python
num_bands = 1
with open('full_dct.%dband.py'%num_bands, 'r') as f:   
    data = eval(f.read())

keys = ["t'","Tcmax","Family","#No"]

cols = [tuple(row[key] for key in keys) for row in data]

for x,y,fam,no in cols:
    ms,clr = ms_and_clr_by_family[fam]
    plt.plot(x,y,ms, color=clr) 
    plt.annotate(no,(x,y))
plt.ylim(0,)
plt.xlabel(keys[0])
plt.ylabel(keys[1])
```
